```python
#Gráfico de barras: top 10 plataformas
top_plataformas = games.groupby("Platform")["Global_Sales"].sum().nlargest(10).reset_index()
top_plataformas
```

```python
#Gráfico de barras: top 10 publishers
top_publishers = games.groupby("Publisher")["Global_Sales"].sum().nlargest(10).reset_index()
top_publishers
```

```python
#Qual gênero vende mais em média
genero_vende_mais = games.groupby("Genre")["Global_Sales"].mean().sort_values(ascending=False)
genero_vende_mais.round(2)
```

```python
#Qual publisher é mais consistente (menor desvio padrão)
contagem1 = games["Publisher"].value_counts()
publishers_validos = contagem1[contagem1 > 50].index
dados_filtrados = games[games["Publisher"].isin(publishers_validos)]


desvio_por_publisher = dados_filtrados.groupby("Publisher")["Global_Sales"].std()
desvio_por_publisher.nsmallest(10).round(2)
```

```python
#Analisando a quantidade de jogos e venda global média das plataforma
jogos_por_plataforma = games.groupby("Platform")["Name"].count().reset_index().sort_values(by="Name", ascending=False)
jogos_por_plataforma
```

```python
mais_baixa = pd.merge(jogos_por_plataforma, media_por_plataforma, on="Platform").sort_values(by=["Name", "Global_Sales"], ascending=[False, False])
mais_baixa.round(2)
```

```python
correlacao = games[["NA_Sales", "EU_Sales", "JP_Sales", "Global_Sales"]].corr()
correlacao.round(2)
```

```python
#Crie um ranking de plataformas por eficiência
eficiencia_plataforma = (
    games
    .groupby("Platform")
    .agg(
        quantidade=("Name", "count"),
        media_vendas=("Global_Sales", "mean"),
        total_vendas=("Global_Sales", "sum")
    )
    .reset_index()
)
```

```python
plataforma_mais_eficiente = eficiencia.sort_values(by="media_vendas", ascending=False)
plataforma_mais_eficiente.round(2)
```
